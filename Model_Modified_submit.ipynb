```python
import pandas as pd
import numpy as np
import joblib  # ใช้สำหรับ save/load model
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.preprocessing import LabelEncoder

# โหลด dataset
df = pd.read_csv('data_input_modified.csv')

# เติม missing
df['person_emp_length'] = df['person_emp_length'].fillna(df['person_emp_length'].median())

# แปลง categorical เป็น numeric
le_home = LabelEncoder()
df['person_home_ownership'] = le_home.fit_transform(df['person_home_ownership'])

le_default = LabelEncoder()
df['cb_person_default_on_file'] = le_default.fit_transform(df['cb_person_default_on_file'])

# แยก X, y
X = df.drop('loan_status', axis=1)
y = df['loan_status']

# แบ่ง train/test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# เทรน model
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

# ประเมิน
y_pred = rf.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('Classification Report:\n', classification_report(y_test, y_pred))

# ====== SAVE MODEL ======
joblib.dump(rf, 'credit_scoring_model.pkl')
joblib.dump(le_home, 'le_home.pkl')
joblib.dump(le_default, 'le_default.pkl')

# ====== LOAD MODEL FOR USE ======
rf_loaded = joblib.load('credit_scoring_model.pkl')
le_home_loaded = joblib.load('le_home.pkl')
le_default_loaded = joblib.load('le_default.pkl')
```
